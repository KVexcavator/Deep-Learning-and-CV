- Глубокое обучение (deep learning) представляет собой качественный сдвиг по сравнению с традиционным машинным обучением. В то время как классические методы машинного обучения в значительной степени опираются на проектирование признаков (feature engineering), методы глубокого обучения спроектированы таким образом, чтобы автоматически выявлять взаимосвязи между входными данными и требуемыми результатами на основе предоставленных примеров.
- Такие библиотеки, как PyTorch, позволяют эффективно создавать и обучать нейросетевые модели для задач глубокого обучения.
- С момента своего появления в начале 2017 года PyTorch стал одним из ведущих фреймворков в сфере глубокого обучения. TensorFlow по-прежнему активно используется в индустрии, а библиотека Hugging Face завоевала большую популярность в качестве высокоуровневой надстройки над различными фреймворками для глубокого обучения.
- В PyTorch приоритет отдается гибкости и скорости работы; библиотека минимизирует когнитивную нагрузку на разработчика и по умолчанию использует режим немедленного (пошагового) выполнения операций.
- В основе PyTorch лежат тензоры — многомерные массивы; библиотека предоставляет обширный набор операций, которые можно выполнять с этими структурами.
