## часть 4: Broadcasting
Трансляция — это механизм, упрощающий сложные тензорные операции, позволяющий эффективно выполнять поэлементные вычисления между тензорами различной формы. Во время трансляции
размеры сравниваются поэлементно справа налево. Если размеры совместимы, тензоры могут быть переданы вместе. 
Совместимые размеры — это
- Одинаковые размерности.
- Одна из размерностей равна 1.

Пример со скалярами, где скалярное значение применяется ко всем элементам тензора:
    Tensor X:       Scalar Y:
[[ 1 2 3 ]]     +   10    
Результат (скалярное сложение с трансляцией):
[[ 11 12 13 ]]

При умножении тензора размерности 1 × 3 на тензор размерности 3 × 1 мы сопоставляем их размерности и видим, что они совместимы. Первая размерность первого тензора равна 1, а вторая размерность второго тензора также равна 1. Следовательно, мы можем применить к этим тензорам механизм расширения (broadcasting), в результате чего получим тензор размерности 3 × 3:
3 x 1
1 x 3
------
3 x 3

Tensor A:           Tensor B:
[[ 1 2 3 ]]         [[10]
                *    [20]
                     [30]]

Результат (широковещательное умножение):
[[ 10 20 30 ]
 [ 20 40 60 ]
 [ 30 60 90 ]]

Если тензоры имеют различающиеся размерности, они выравниваются по правому краю. Если размерности совпадают, операция выполняется поэлементно. А если одна из размерностей равна 1, можно считать, что значения вдоль этой размерности как бы дублируются, чтобы соответствовать большей размерности. 

Пример с тензорами размерностей 2 × 1 × 3 и 2 × 3:
    2 x 3
2 x 1 x 3
----------
2 x 2 x 3

Matrix A:       Tensor B:
[[ 1 2 3 ]      [[[10 20 30]]
            +
[ 4 5 6 ]]      [[40 50 60]]]

Результат (сложение тензоров с использованием механизма расширения размерностей):

[[[ 11 22 33 ]
  [ 44 55 66 ]]
 [[ 41 52 63 ]
  [ 44 55 66 ]]]

## часть 5: Именованные тензоры

Допустим, есть трехмерный тензор `img_t` и мы хотим преобразовать его в полутоновое изображение. Можно воспользоваться стандартными весовыми коэффициентами для цветовых каналов, чтобы получить единое значение яркости:

````python
img_t = torch.randn(3, 5, 5) # shape [channels, rows, columns]
weights = torch.tensor([0.2126, 0.7152, 0.0722])
````

Часто также требуется, чтобы наш код был универсальным — например, чтобы он мог работать как с полутоновыми изображениями (представленными в виде двумерных тензоров с измерениями высоты и ширины), так и с цветными изображениями (добавляя третье измерение — канал, как в формате RGB), а также чтобы он мог обрабатывать не одно изображение, а целый их набор (батч). Допустим, что у нас имеется батч из двух изображений:

````python
batch_t = torch.randn(2, 3, 5, 5) # shape [batch, channels, rows, columns]
````

Иногда RGB-каналы располагаются в измерении 0, а иногда — в измерении 1. Однако можно обобщить это, ведя отсчет с конца: они всегда находятся в измерении -3 (третьем с конца). 
Таким образом, простое (невзвешенное) среднее значение можно записать следующим образом:

````python
img_gray_naive = img_t.mean(-3)
batch_gray_naive = batch_t.mean(-3)
img_gray_naive.shape, batch_gray_naive.shape
````

Тензор `batch_t` формы (2, 3, 5, 5) умножается на тензор `unsqueezed_weights` формы (3, 1, 1), в результате чего получается тензор формы (2, 3, 5, 5). После этого можно просуммировать значения по третьей с конца размерности (соответствующей трем каналам):

````python
unsqueezed_weights = weights.unsqueeze(-1).unsqueeze_(-1)
img_weights = img_t * unsqueezed_weights
batch_weights = batch_t * unsqueezed_weights
img_gray_weighted = img_weights.sum(-3)
batch_gray_weighted = batch_weights.sum(-3)
batch_weights.shape, batch_t.shape, unsqueezed_weights.shape
````

Размерности тензоров не всегда очевидны — чтобы их определить, приходится прослеживать всю цепочку вычислений. Такой подход чреват ошибками, особенно если места создания тензоров и их использования сильно удалены друг от друга в коде.

### часть 6: Типы элементов тензора

Библиотеки для анализа данных опираются на NumPy или вводят специализированные структуры данных — например, тензоры PyTorch. Они обеспечивают эффективную низкоуровневую реализацию числовых структур и операций над ними, предоставляя при этом удобный высокоуровневый программный интерфейс (API). Чтобы такая структура могла функционировать, все элементы тензора должны быть числами одного типа, а библиотека PyTorch должна отслеживать этот числовой тип.

Аргумент `dtype` в конструкторах тензоров (таких как `tensor`, `zeros` и `ones`) определяет тип числовых данных, которые будут содержаться в тензоре. Тип данных задает возможные значения, которые может принимать тензор (целые числа или числа с плавающей запятой), а также количество байтов, отводимых на каждое значение (а в случае с `uint8` — еще и наличие или отсутствие знака).

- torch.float32 или torch.float — 32-битные числа с плавающей точкой
- torch.float64 или torch.double — 64-битные числа с плавающей точкой (двойной точности)
- torch.complex64 или torch.cfloat — 64-битные комплексные числа
- torch.complex128 или torch.cdouble — 128-битные комплексные числа
- torch.float16 или torch.half — 16-битные числа с плавающей точкой (половинной точности)
- torch.bfloat16 — 16-битные числа с плавающей точкой формата «brain float» (половинной точности)
- torch.int8 — 8-битные целые числа со знаком
- torch.uint8 — 8-битные целые числа без знака
- torch.int16 или torch.short — 16-битные целые числа со знаком
- torch.int32 или torch.int — 32-битные целые числа со знаком
- torch.int64 или torch.long — 64-битные целые числа со знаком
- torch.bool — логический тип (булевы значения)

## Часть 8: Тензоры: взгляд на организацию памяти

Значения в тензорах размещаются в непрерывных блоках памяти, управление которыми осуществляется с помощью объектов `Storage` (хранилищ). Хранилище представляет собой одномерный массив числовых данных — непрерывный блок памяти, содержащий числа определенного типа. Экземпляр тензора в PyTorch — это своего рода «представление» (view) такого хранилища, позволяющее обращаться к его данным с использованием смещения и шагов (strides) для каждого измерения.
Несколько тензоров могут ссылаться на одно и то же хранилище, даже если они обращаются к содержащимся в нем данным по-разному.

```text
Хранилище (storage) — одна полка в памяти:

 индекс:   0   1   2   3   4   5
         ┌───┬───┬───┬───┬───┬───┐
         │ 4 │ 1 │ 5 │ 3 │ 2 │ 1 │
         └───┴───┴───┴───┴───┴───┘
               ▲               ▲
               │               │
     ┌─────────┘               └─────────┐
     │                                   │
 points                         points_t = points.t()
 shape (3, 2), stride (2, 1)    shape (2, 3), stride (1, 2)
 ┌───┬───┐                      ┌───┬───┬───┐
 │ 4 │ 1 │                      │ 4 │ 5 │ 2 │
 ├───┼───┤                      ├───┼───┼───┤
 │ 5 │ 3 │                      │ 1 │ 3 │ 1 │
 ├───┼───┤                      └───┴───┴───┘
 │ 2 │ 1 │
 └───┴───┘

Числа не копируются: оба тензора смотрят в одно хранилище,
отличаются только shape и stride.
```

```text
Полка в памяти (storage):

 место:   0     1     2     3     4     5
        ┌─────┬─────┬─────┬─────┬─────┬─────┐
        │ [4] │  1  │ [5] │  3  │ [2] │  1  │
        └─────┴─────┴─────┴─────┴─────┴─────┘
           └──── +2 ───┘└──── +2 ────┘
          [4], [5], [2] лежат через одно место


points, stride (2, 1)             points_t = points.t(), stride (1, 2)

       вправо: +1                        вправо: +2
         ──────►                   ─────────────────►
      ┌─────┬─────┐               ┌─────┬─────┬─────┐
   │  │ [4] │  1  │            │  │ [4] │ [5] │ [2] │
вниз  ├─────┼─────┤         вниз  ├─────┼─────┼─────┤
 +2│  │ [5] │  3  │          +1│  │  1  │  3  │  1  │
   │  ├─────┼─────┤            ▼  └─────┴─────┴─────┘
   ▼  │ [2] │  1  │
      └─────┴─────┘

[4], [5], [2] в points стоят столбиком  → прыжок +2 идёт вниз
[4], [5], [2] в points_t стоят строкой  → прыжок +2 идёт вправо

stride = (сколько мест прыгнуть вниз, сколько мест прыгнуть вправо)
```

````python
# Индексация в хранилище
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points.storage()
````

Хотя тензор(выше) и позиционируется как структура с тремя строками и двумя столбцами, на самом деле в памяти он представлен в виде непрерывного массива из 6 элементов. Иными словами, тензор просто «знает», как преобразовать пару индексов в конкретное местоположение в этом массиве.

````python
points_storage = points.storage()
points_storage[0]
````

````python
points.storage()[1]
````

````python
# Невозможно обращаться к хранилищу двумерного тензора, используя два индекса. Структура хранилища всегда одномерна, независимо от размерности тензоров, которые могут на него ссылаться.
# Изменение значения в хранилище приводит к изменению содержимого ссылающегося на него тензора:
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points_storage = points.storage()
points_storage[0] = 2.0
points
````

Существует небольшое число операций, представленных исключительно в виде методов объекта Tensor. Их можно распознать по символу подчеркивания в конце названия (например, `zero_`); это указывает на то, что метод выполняет операцию «на месте» (in-place) — то есть изменяет исходные данные, а не создает и не возвращает новый тензор. 

````python
# zero_ обнуляет все элементы входного тензора
a = torch.ones(3, 2)
a.zero_()
a
````

Непрерывные тензоры

Некоторые операции с тензорами в PyTorch работают только с непрерывными (contiguous) тензорами — например, операция `view`. В таких случаях PyTorch выдает понятное сообщение об ошибке и требует явного вызова метода `contiguous`.

````python
points.is_contiguous()
````

````python
points_t.is_contiguous()
````


С помощью метода `contiguous` можно получить новый тензор с непрерывным размещением данных в памяти (contiguous tensor) из тензора, у которого такое размещение отсутствует. Содержимое тензора останется прежним, однако изменится шаг (stride), поскольку, по сути, происходит переупорядочивание значений в новой области памяти:

````python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points_t = points.t()
points_t
````

### Часть 10: Перенос тензоров на GPU

Тензоры PyTorch также могут храниться на процессоре другого типа — графическом процессоре (GPU).

Поддерживаются как графические процессоры Nvidia с технологией CUDA, так и платформа AMD ROCm. Обе эти платформы имеют встроенную поддержку, и их можно установить непосредственно с помощью pip или conda. Для пользователей Mac на базе Apple Silicon в PyTorch предусмотрена поддержка Metal Performance Shaders (MPS). Поддержка тензорных процессоров (TPU) от Google реализована через отдельный пакет torch_xla (https://github.com/pytorch/xla). Аналогичным образом, поддержка устройств XPU от Intel доступна через пакет intel_extension_for_pytorch (https://github.com/intel/intel-extension-for-pytorch).

> ⚠️ **Не на моём железе:** нет видеокарты NVIDIA (CUDA), PyTorch собран `+cpu`.
> Код книги с `'cuda'` упадёт с ошибкой `Torch not compiled with CUDA enabled`.
> Запускать с GPU — в Google Colab (Runtime → Change runtime type → T4 GPU).

Кроме `dtype`, у тензора есть **`device`** — где лежат его данные: в обычной памяти (`cpu`) или в памяти видеокарты (`cuda`). На GPU математические операции выполняются намного быстрее. API у CPU- и GPU-тензоров одинаковый, поэтому код почти не зависит от того, где идут вычисления.

```python
# создать тензор сразу на GPU
points_gpu = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]], device='cuda')

# скопировать тензор с CPU на GPU
points_gpu = points.to(device='cuda')

# если GPU несколько — указать номер (с нуля)
points_gpu = points.to(device='cuda:0')

# умножение на CPU
points = 2 * points
# умножение на GPU: тензор копируется на GPU, результат тоже остаётся на GPU
points_gpu = 2 * points.to(device='cuda')

# сложение тоже идёт на GPU, данные на CPU не возвращаются
# (пока мы не выведем или не прочитаем результат)
points_gpu = points_gpu + 4

# вернуть тензор на CPU
points_cpu = points_gpu.to(device='cpu')

# короткие формы
points_gpu = points.cuda()
points_gpu = points.cuda(0)
points_cpu = points_gpu.cpu()
```

Через `.to()` можно одновременно поменять и устройство, и тип: `points.to(device='cuda', dtype=torch.double)`.

Как пишут на практике (работает везде)
```python
device = "cuda" if torch.cuda.is_available() else "cpu"
points_dev = points.to(device)       # на GPU, если есть, иначе остаётся на CPU
points_dev = 2 * points_dev + 4      # считается там, где лежит тензор
points_cpu = points_dev.cpu()        # вернуть на CPU (например, для numpy)
```

Главное
1. У тензора есть `device` — `cpu` или `cuda`.
2. `.to(device)` переносит тензор, `.cpu()` возвращает обратно.
3. **Тензоры в одной операции должны быть на одном устройстве**, иначе ошибка
   `Expected all tensors to be on the same device`.


### Глава 11: Взаимодействие с NumPy

Тензоры PyTorch можно преобразовывать в массивы NumPy и наоборот и получить доступ к функционалу Python и NumPy. Такая возможность взаимодействия с массивами NumPy без копирования данных (zero-copy) обеспечивается тем, что система хранения данных работает с протоколом буферов Python (Python buffer protocol)

````python
points = torch.ones(3, 4)
points_np = points.numpy()
points_np
````

Если тензор размещён на GPU, PyTorch создаст копию содержимого этого тензора в виде массива NumPy, размещённого в CPU.

````python
# назад в Pytorch
points = torch.from_numpy(points_np)
points
````

### Глава 12: Обобщенные тензоры — это тоже тензоры
PyTorch обеспечивает вызов соответствующих вычислительных функций независимо от того, где находится тензор — на центральном (CPU) или графическом (GPU) процессоре. Это достигается благодаря механизму диспетчеризации. 
Существуют другие виды тензоров: одни предназначены для определенных классов аппаратного обеспечения (например, для тензорных процессоров Google — TPU), а другие используют иные способы представления данных, отличные от привычных нам плотных массивов (dense arrays). Так, разреженные тензоры (sparse tensors) хранят только ненулевые элементы вместе с информацией об их индексах.
Иногда стандартные тензоры называют «плотными» (dense) или «тензорами со страйдами» (strided), чтобы отличать их от тензоров с иной организацией памяти.
Количество разновидностей тензоров постоянно растет.

Сохранение в HDF5 с помощью h5py

> ⏭ **Пропущено, не запускал** — нужно редко, только для обмена данными с другими системами.
> Для своих проектов на PyTorch хватает `torch.save` / `torch.load`.
> Если понадобится: `pip install h5py`

**Зачем:** сохранять тензоры так, чтобы их могли прочитать другие программы и библиотеки. Чаще нужно при встраивании PyTorch в уже существующую систему; новым проектам — редко.

**HDF5** — переносимый, широко поддерживаемый формат для хранения многомерных массивов, организованных во вложенный словарь «ключ → значение». В Python с ним работает библиотека **h5py**, которая принимает и возвращает данные в виде массивов NumPy.

Сохранить
Тензор переводим в NumPy (бесплатно, без копирования) и сохраняем под ключом `'coords'`:
```python
import h5py
f = h5py.File('./data/p1ch3/ourpoints.hdf5', 'w')
dset = f.create_dataset('coords', data=points.numpy())
f.close()
```
Ключей в файле может быть много, в том числе вложенных.

Прочитать часть данных прямо с диска
Главная фишка HDF5: можно взять только нужный кусок, не загружая весь файл в память.
```python
f = h5py.File('./data/p1ch3/ourpoints.hdf5', 'r')
dset = f['coords']
last_points = dset[-2:]      # с диска читаются только две последние точки
```
При открытии файла данные не загружаются — только в момент обращения к срезу. Возвращается объект, который ведёт себя как массив NumPy.

Сразу в тензор
```python
last_points = torch.from_numpy(dset[-2:])   # данные копируются в тензор
f.close()
```
После `f.close()` обращаться к `dset` нельзя — будет ошибка. Поэтому сначала забираем данные в тензор, потом закрываем файл.
