3 Часто используемые функции

```python
import numpy as np
```

File I/O

Единичная матрица — это квадратная матрица, у которой на главной диагонали стоят единицы, а все остальные элементы равны нулю.

```python
i2 = np.eye(2)
print(i2)
```

```python
# Сохранить
np.savetxt("eye.txt", i2)
```

Comma-seperated value files

Функция `loadtxt()` позволяет удобно считывать CSV-файлы, разделять их на отдельные поля и загружать данные в массивы NumPy.

Например, жанные хранятся в файле `data.csv`:
AAPL,28-01-2011, ,344.17,344.4,333.53,336.1,21144800
Нас интересуют только цена закрытия (close price) и объем торгов (volume).
В качестве разделителя (`delimiter`) мы указали запятую (`,`).
Параметр `usecols` задан в виде кортежа, что позволяет выбрать седьмое и восьмое поля, соответствующие цене закрытия и объему торгов. 
Аргумент `unpack` установлен в значение `True`: это означает, что данные будут распакованы и присвоены переменным `c` и `v`, в которых будут храниться цена закрытия и объем торгов соответственно.

```python
с, v = np.loadtxt('data.csv', delimiter=',', usecols=(6,7), unpack=True)
с
```

Средневзвешенная по объему цена
Volume Weighted Average Price (VWAP)

```python
vwap = np.average(c, weights=v)
print("VWAP =", vwap)
```

Функция mean() вычисляет среднее арифметическое массива. Она суммирует значения в массиве a и делит полученную сумму на количество элементов n.

```python
print("mean= ", np.mean(c))
```

Средневзвешенная по времени цена
time-weighted average price (TWAP)

```python
t = np.arange(len(c))
print("twap =", np.average(c, weights=t))
```

поиск наибольшего и наименьшего значений

```python
h,l=np.loadtxt('data.csv', delimiter=',', usecols=(4,5),
unpack=True)
print("highest =", np.max(h))
print("lowest =", np.min(l))
```

Функция ptp() возвращает разность между максимальным и минимальным значениями массива.

```python
print("Spread high price", np.ptp(h))
print("Spread low price", np.ptp(l))
```

Поиск медианы.
Одна половина данных находится ниже медианы, а другая — выше неё. Например, если у нас есть значения 1, 2, 3, 4 и 5, то медианой будет 3, поскольку это число находится посередине.

```python
c = np.loadtxt('data.csv', delimiter=',', usecols=(6,), unpack=True)
print("median =", np.median(c))
```

Дисперсия(Variance) показывает, насколько сильно варьируется переменная. Она также указывает на уровень риска инвестиций: если цена акции меняется слишком резко, это неизбежно может создать для нас проблемы.
Рассчёт дисперсии цены закрытия:

```python
print("variance =", np.var(c))
```

std — это корень из дисперсии
То есть цена обычно гуляет примерно на ±7 вокруг среднего 351.

```python
print("std =", np.std(c))
```

max отвечает на вопрос «какое самое большое число», а argmax — «где оно стоит» (номер элемента):

```python
i = np.argmax(c)
print("номер дня:", i)
print("цена в этот день:", c[i])
```

clip
Всё, что ниже 340, превращается в 340, всё, что выше 355, — в 355, а остальное не меняется:

```python
print(np.clip(c, 340, 355))
```
