Срезы и индексация многомерных массивов
Класс ndarray поддерживает создание срезов по нескольким измерениям. Для удобства можно обращаться к нескольким измерениям одновременно, используя многоточие.

Пример
Массив b содержит 24 элемента со значениями от 0 до 23; мы преобразовали его в трехмерный массив размерности 2×3×4.
Это можно представить как двухэтажное здание, на каждом этаже которого расположено 12 комнат, организованных в 3 ряда по 4 комнаты в каждом.

````python
b = np.arange(24).reshape(2, 3, 4)
print(b.shape)
print(b)
# выбрать отдельную комнату, используя три её координаты: этаж, столбец и ряд
print(b[0, 0, 0])
# первый столбец и строка
print(b[: 0, 0 ])
# выбрать первый этаж
print(b[0])
# или
print(b[0, :, :])
# или
print(b[0, ...])
# второй ряд на пeрвом этаже
print(b[0, 1])
# там же выбрать какжый второй элемент
print(b[0, 1, ::2])
# Выберите все помещения на обоих этажах, которые находятся во втором столбце
print(b[... , 1])
# все номера во втором ряду, независимо от этажа и столбца
print(b[:, 1])
# помещения на первом этаже, второй столбец
print(b[0, :, 1])

# использование негативных индексов
# первый этаж, последний столбец
print(b[0, :, -1])
# номера на первом этаже, последний столбец — в обратном порядке
print(b[0, :: -1, -1])
# тут же каждый второй элемент
print(b[0, ::2, -1])

# Команда, которая разворачивает одномерный массив, помещает верхний этаж вслед за первым этажом
print(b[:: -1])
````

При «уплощении» многомерных массивов NumPy получается одномерный массив, содержащий те же данные.

````python
print(b)
print(b.ravel())
print(b.flatten())
````

помимо функции reshape(), мы также можем задать форму напрямую с помощью кортежа

````python
b.shape = (6, 4)
print(b)
````

Метод resize() работает так же, как функция reshape(), но при этом изменяет сам массив, к которому он применяется:

**Функции изменения формы массива:**

| Функция | Описание |
|---|---|
| `ravel()` | Возвращает одномерный массив с теми же данными. Копию создаёт не всегда: по возможности возвращает *представление* (view) исходного массива |
| `flatten()` | Метод `ndarray`: «расплющивает» массив в одномерный и **всегда** возвращает копию |
| `reshape()` | Возвращает массив с новой формой; исходный массив не меняется |
| `resize()` | Меняет форму и размер массива. `np.resize(a, ...)` возвращает новый массив и при нехватке элементов заполняет его повторами исходных; метод `a.resize(...)` меняет сам массив на месте и дополняет его нулями |

````python
b = 2 * a
b
````

**Функции объединения массивов:**

| Функция | Описание |
|---|---|
| `vstack()` | Объединяет массивы по вертикали |
| `dstack()` | Объединяет массивы по глубине, вдоль третьей оси |

| `concatenate()` | Объединяет список или кортеж массивов |

Комплексные числа на кошечках 🐈

### Кошка в комнате

Представим кошку на полу. Чтобы сказать, где она сидит, одного числа мало, нужно два:
- сколько шагов **вперёд** от двери,
- сколько шагов **вбок**.

Комплексное число — это и есть такой **адрес кошки из двух частей**:

```
3 + 2j   →   3 шага вперёд (real) и 2 шага вбок (imag)
```

Буква `j` просто помечает, какое число относится к шагам вбок, чтобы их не перепутать с шагами вперёд.

### Массив = несколько кошек

```python
c = np.array([1 + 1j, 3 + 2j])
```

Две кошки:
- Мурка: 1 вперёд, 1 вбок
- Барсик: 3 вперёд, 2 вбок

```python
c.real   # [1., 3.]  — сколько вперёд у Мурки и у Барсика
c.imag   # [1., 2.]  — сколько вбок у Мурки и у Барсика
```

`.real` и `.imag` — это как спросить у всех кошек сразу: «а вперёд вы на сколько?», «а вбок?»

### А при чём тут странное «j² = −1»?

Умножить на `j` — это **повернуть кошку на 90° влево**:

```
кошка смотрит вперёд       →  1
× j  → повернулась влево   →  j
× j  → ещё раз (уже назад) →  -1
```

Два поворота по 90° — и кошка смотрит **назад**, то есть j × j = −1. В этом вся «магия» мнимой единицы: это не загадочное число, а **поворот**.

### А `np.abs()`?

```python
np.abs(3 + 4j)   # 5.0
```

Это **расстояние от двери до кошки по прямой**, если идти напрямик, а не «вперёд, потом вбок». По теореме Пифагора: √(3² + 4²) = 5.

---

**Итого:** комплексное число — координаты кошки на полу: `real` — вперёд, `imag` — вбок. Умножение на `j` поворачивает кошку, `abs` даёт расстояние до неё. 🐾

## 🗺 Карта NumPy

### Одна вещь — массив
| Свойство | Что это | Пример |
|---|---|---|
| данные | сами числа | `[0, 1, 2, 3]` |
| `shape` | форма | `(2, 12)`, `(3, 3, 3)` |
| `dtype` | тип чисел | `int64`, `float32`, `uint8` |

### Четыре действия
| Действие | Вопрос | Функции |
|---|---|---|
| **1. Создать** | откуда взять массив? | `np.array`, `np.arange`, `np.loadtxt`, `np.eye`, `np.zeros`, `np.ones` |
| **2. Выбрать** | какую часть взять? | `a[0, 1]`, `a[::2]`, `a[0, ...]`, `a.flat`, `a[a > 5]` |
| **3. Переформить** | как переложить те же числа? | `reshape`, `ravel`, `flatten`, `T`, `vstack`, `hstack`, `split` |
| **4. Посчитать** | что из них получить? | см. ниже |

### Посчитать — два вида
| Вид | Что происходит с формой | Функции |
|---|---|---|
| **поэлементно** | форма та же | `a + b`, `a * 2`, `a ** 2`, `a > 5`, `clip` |
| **сводка** | чисел становится меньше | `sum`, `mean`, `median`, `std`, `var`, `max`, `min`, `argmax`, `ptp` |

### Правила
- Новая функция → спроси: **к какому из 4 действий она относится?**
- Перед запуском ячейки → **угадай результат** (хотя бы `shape`).
- Финансовые примеры книги — это сюжет, запоминать нужно только функции.
