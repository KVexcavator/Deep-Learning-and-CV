### 1. Создайть виртуальное окружение
```powershell
python -m venv .venv
```
- `python -m venv` запускает встроенный модуль venv.
- `.venv` — имя папки, куда ляжет отдельная копия Python.
- Команда ничего не пишет, это нормально. Проверка: `ls -Force`, должна появиться папка `.venv`.

### 2. Поставить пакеты в это окружение
```powershell
.\.venv\Scripts\python.exe -m pip install numpy ipykernel matplotlib
```
- Здесь важно, что вызывается python **из `.venv`**, а не системный, поэтому пакеты ставятся именно туда.
- `numpy` — то, что изучаем; `ipykernel` нужен, чтобы VS Code мог запускать блокноты на этом Python; `matplotlib` пригодится для графиков.
- В конце должно быть `Successfully installed ...`.

Проверить:
```powershell
.\.venv\Scripts\python.exe -c "import numpy; print(numpy.__version__)"
```

### 6 Создание вектора на чистом Python

````python
def pythonsum(n):
    a = range(n)
    b = range(n)
    c = []

    for _ in range(len(a)):
        a[i] = i ** 2
        b[i] = i ** 3
        c.append(a[i] + b[i])

    return c
````

### 7 Cоздание вектора с NumPy

````python

import numpy as np

def numpyone(n):
    a = np.arange(n) ** 2
    return a

print("NumpyOne: ", numpyone(3))

def numpysum(n):
    a = np.arange(n) ** 2
    b = np.arange(n) ** 3
    c = a + b
    return c

print("NumpySum: ", numpysum(3))
````
