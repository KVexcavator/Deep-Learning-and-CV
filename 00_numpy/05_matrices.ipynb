5. Матрицы и универсальные функции (ufuncs).

```python
import numpy as np
```

Матрицы
В NumPy матрицы являются подклассами `ndarray`. Их можно создавать, используя специальный строковый формат. Матрицы всегда двумерны. Операция умножения матриц отличается от обычного умножения в NumPy. То же самое касается и операции возведения в степень.

Cоздание матриц

Функция `np.asmatrix()` не создаёт копию,
если входные данные уже являются матрицей (`matrix`) или массивом (`ndarray`).
Вызов этой функции равносилен вызову `np.matrix(data, copy=False)`.


```python
A = np.asmatrix('1 2 3; 4 5 6; 7 8 9')
print("Создание из строки: ", A)
```

```python
print("transpose A", A.T)
```

```python
print("Создание из массива", np.asmatrix(np.arange(9).reshape(3, 3)))
```

Cоздание матриц из других матриц

```python
# создание единичной матрицы размера 2×2:
A = np.eye(2)
print("A", A)
```

```python
# Создание ещё одной матрицы, подобной A, и умноженной на 2:
B = 2 * A
print("B", B)
```

```python
# Создание составной матрицы на основе строки
print("Составная матрица ", np.bmat('A B; A B'))
```

Универсальные функции (ufuncs) принимают на вход набор скалярных величин и выдают на выходе также набор скалярных величин. По сути, это объекты Python, инкапсулирующие поведение той или иной функции. Как правило, такие функции соответствуют математическим операциям — сложению, вычитанию, делению, умножению и т. д. В целом, универсальные функции работают быстрее благодаря специальным оптимизациям и выполнению на низкоуровневом (нативном) уровне.

```python
# создаем какуюто функцию на Python
def ultimate_answer(a):
    # результат, состоящий из одних нулей и имеющий ту же форму, что и a
    result = np.zeros_like(a)
    result.flat = 42
    return result

# 1 — количество входных параметров, за которым следует 1 — количество выходных параметров:
ufunc = np.frompyfunc(ultimate_answer, 1, 1)
print("Ответ ", ufunc(np.arange(4)))
```

```python
# то же с двухуровневым массавом
print("Ответ ", ufunc(np.arange(4).reshape(2,2)))
```

Методы универсальных функций

```python
a = np.arange(9)
a
```

```python
# Универсальная функция выполняет рекурсивное сокращение входного массива вдоль заданной оси, последовательно обрабатывая его элементы. В случае функции add() результат такого сокращения аналогичен вычислению суммы элементов массива.
print("Reduce", np.add.reduce(a))
```

```python
# Метод accumulate() сохраняет промежуточные результаты в массиве и возвращает его
print("Accumulate", np.add.accumulate(a))
```

Метод reduceat() требует в качестве аргументов входной массив и список индексов:

```python
print("Reduceat", np.add.reduceat(a, [0, 5, 2, 7]))
```

Пошаговое оьяснение:

```python
# Первый шаг касается индексов 0 и 5. Этот шаг приводит к операции свёртки (reduce) элементов массива, расположенных между индексами 0 и 5:
print("Reduceat step 1: ", np.add.reduce(a[0:5]))
```

```python
# Второй шаг касается индексов 5 и 2. Поскольку 2 меньше 5, возвращается элемент массива с индексом 5:
print("Reduceat step 2: ", a[5])
```

```python
# Третий шаг касается индексов 2 и 7. Этот шаг приводит к операции свёртки элементов массива, расположенных между индексами 2 и 7:
print("Reduceat step 3: ", np.add.reduce(a[2:7]))
```

```python
# Четвертый шаг относится к индексу 7. Этот шаг приводит к выполнению операции свертки (reduce) для элементов массива, начиная с индекса 7 и до конца массива:
print("Reduceat step 4: ", np.add.reduce(a[7:]))
```

Метод outer() возвращает массив, ранг которого равен сумме рангов двух входных массивов. Этот метод применяется ко всем возможным парам элементов входных массивов.

```python
other = np.arange(3)
other
```

```python
print("Outer", np.add.outer(other, a))
```

Операция взятия остатка
Вычислить остаток от деления можно с помощью функций `mod()`, `remainder()` и `fmod()` из библиотеки NumPy, а также используя оператор `%`. Основное различие между этими функциями заключается в том, как они обрабатывают отрицательные числа. При этом функция `fmod()` выделяется на фоне остальных.

```python
a = np.arange(-4, 4)
a
```

```python
print("Remainder", np.remainder(a, 2))
```

```python
print("Mod", np.mod(a, 2))
```

```python
print("% operator", a % 2)
```

```python
print("Fmod", np.fmod(a, 2))
```

Побитовые функции и функции сравнения
Побитовые функции работают с битами целых чисел или массивов целых чисел, являясь универсальными функциями. 
Операторы ^, &, |, <<, >> и другие имеют свои аналоги в NumPy. 
То же самое относится и к операторам сравнения, таким как <, >, == и прочим.

```python
x = np.arange(-9, 9)
x
```

```python
y = -x
y
```


**Таблица истинности XOR (исключающее ИЛИ):**

| Вход 1 | Вход 2 | XOR |
|---|---|---|
| `True` | `True` | `False` |
| `False` | `True` | `True` |
| `True` | `False` | `True` |
| `False` | `False` | `False` |


```python
# Оператор ^ соответствует функции bitwise_xor(), а оператор < — функции less():
print("Sign different?", (x ^ y) < 0)
print("Sign different?", np.less(np.bitwise_xor(x, y), 0))
```

**Таблица истинности AND (логическое И):**

| Вход 1 | Вход 2 | AND |
|---|---|---|
| `True` | `True` | `True` |
| `False` | `True` | `False` |
| `True` | `False` | `False` |
| `False` | `False` | `False` |

```python
# В NumPy аналогом оператора & является функция bitwise_and(), а аналогом оператора == — универсальная функция equal():
print("Power of 2?\n", x, "\n", (x & (x - 1)) == 0)
print("Power of 2?\n", x, "\n", np.equal(np.bitwise_and(x, (x - 1)), 0))
```

```python
# В библиотеке NumPy аналогом оператора << является универсальная функция left_shift()
print("Modulus 4\n", x, "\n", x & ((1 << 2) - 1))
print("Modulus 4\n", x, "\n", np.bitwise_and(x,
np.left_shift(1, 2) - 1))
```
